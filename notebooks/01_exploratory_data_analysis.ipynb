# 01. Exploratory Data Analysis (EDA) - SaaS Churn & Revenue Intelligence
**Objective**: Conduct initial data inspection, analyze distributions, detect churn correlations, and understand customer revenue metrics.

```python
import pandas as pd
import numpy as np
import os

# Load processed dataset
data_path = '../data/processed/'
df_customers = pd.read_csv(os.path.join(data_path, 'Dim_Customers.csv'))
df_subs = pd.read_csv(os.path.join(data_path, 'Fact_Subscriptions.csv'))
df_health = pd.read_csv(os.path.join(data_path, 'Fact_Customer_Health_Score.csv'))

print(f"Total Customers: {len(df_customers):,}")
print(f"Total Subscriptions: {len(df_subs):,}")
df_subs.head()
```

### 1. Key Metrics & Overall Churn Rate

```python
total_mrr = df_subs[df_subs['subscription_status'] == 'Active']['monthly_recurring_revenue'].sum()
total_arr = total_mrr * 12
churn_rate = (df_subs['is_churned'].sum() / len(df_subs)) * 100
avg_ltv = df_subs['customer_lifetime_value_ltv'].mean()

print(f"Active MRR: ${total_mrr:,.2f}")
print(f"Active ARR: ${total_arr:,.2f}")
print(f"All-Time Churn Rate: {churn_rate:.2f}%")
print(f"Average LTV: ${avg_ltv:,.2f}")
```

### 2. Churn Breakdown by Billing Cycle & Plan Tier

```python
churn_by_plan = df_subs.groupby('plan_name').agg(
    total_users=('customer_id', 'count'),
    churned_users=('is_churned', 'sum'),
    avg_ltv=('customer_lifetime_value_ltv', 'mean')
).reset_index()
churn_by_plan['churn_rate_pct'] = (churn_by_plan['churned_users'] / churn_by_plan['total_users']) * 100
churn_by_plan.sort_values(by='churn_rate_pct', ascending=False)
```
