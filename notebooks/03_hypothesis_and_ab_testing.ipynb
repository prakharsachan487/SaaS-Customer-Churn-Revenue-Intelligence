# 03. Hypothesis Testing & A/B Retention Experimentation
**Objective**: Conduct statistical hypothesis testing on churn drivers (Contract Type, Support Escalations, Onboarding).

```python
import pandas as pd
import scipy.stats as stats
import os

data_path = '../data/processed/'
df_subs = pd.read_csv(os.path.join(data_path, 'Fact_Subscriptions.csv'))

# Hypothesis 1: Annual Contracts have significantly lower churn rate than Monthly Contracts
monthly_churn = df_subs[df_subs['billing_cycle'] == 'Monthly']['is_churned']
annual_churn = df_subs[df_subs['billing_cycle'] == 'Annual']['is_churned']

t_stat, p_val = stats.ttest_ind(monthly_churn, annual_churn)
print(f"Annual vs Monthly Churn T-Statistic: {t_stat:.4f}, P-Value: {p_val:.4e}")
if p_val < 0.05:
    print(">> REJECT NULL HYPOTHESIS: Annual contracts have a statistically significant lower churn rate!")
```
